## Task-1 Remove duplicate rows and irrelevant columns by implementing a custom data cleaning pipeline that automatically detects and logs redundant data while ensuring no loss of critical information.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
```

```python
duplicate_count = df.duplicated().sum()
print("Duplicate Rows Detected:", duplicate_count)

df = df.drop_duplicates().reset_index(drop=True)
print("Shape After Removing Duplicates:", df.shape)
```

```python
irrelevant_cols = [col for col in df.columns if df[col].nunique() == len(df) or df[col].nunique() <= 1]
print("Irrelevant Columns Detected:", irrelevant_cols)

df = df.drop(columns=irrelevant_cols)
print("Shape After Removing Irrelevant Columns:", df.shape)
```

## Task-2 Standardize column names and correct data types by implementing a dynamic parser that automatically detects and converts incorrect formats, ensuring consistency across the entire dataset without introducing errors.

```python
df.columns = [col.strip().lower().replace(" ", "_") for col in df.columns]
print("Standardized Column Names:")
print(df.columns.tolist())
```

```python
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].astype(str).str.strip().replace("nan", np.nan)
    else:
        df[col] = pd.to_numeric(df[col], errors="coerce")

df["launch_year"] = df["launch_year"].astype("int64")
df["customer_rating"] = df["customer_rating"].astype("int64")
df["recommended_ind"] = df["recommended_ind"].astype("int64")

print(df.dtypes)
```

## Task-3 Deal with outliers in numerical columns by applying multiple detection methods (IQR, Z-score, DBSCAN clustering) and deciding case-by-case whether to cap, transform, or remove extreme values based on domain-specific insights.

```python
num_cols = ["mrp", "monthly_units_sold", "positive_feedback_count"]

for col in num_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    iqr_outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    
    z_scores = np.abs((df[col] - df[col].mean()) / df[col].std())
    z_outliers = (z_scores > 3).sum()
    
    print(f"{col} -> IQR Outliers: {iqr_outliers} | Z-Score Outliers: {z_outliers}")
```

```python
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df[num_cols].fillna(df[num_cols].median()))

dbscan = DBSCAN(eps=1.5, min_samples=10)
clusters = dbscan.fit_predict(scaled_features)

print("DBSCAN Outliers Detected:", (clusters == -1).sum())
```

```python
for col in ["mrp", "monthly_units_sold"]:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    df[col] = df[col].clip(lower=lower, upper=upper)

df["positive_feedback_count"] = np.log1p(df["positive_feedback_count"])

df = df[clusters != -1].reset_index(drop=True)
print("Final Shape After Outlier Treatment:", df.shape)
```

```python
df.to_csv("cleaned_day1.csv", index=False)
print("Saved cleaned dataset to cleaned_day1.csv")
```

## Normalize or scale numerical features for consistency by testing multiple scaling techniques (Min-Max, StandardScaler, RobustScaler, Power Transform) and selecting the most suitable approach for preserving data distributions while optimizing model perform

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, PowerTransformer
from scipy.stats import entropy

sns.set_theme(style="whitegrid")
```

```python
num_cols = [
    "customer_age", "mrp", "discount_pct", "selling_price",
    "material_quality_score", "review_word_count", "sentiment_score",
    "social_media_trend_score", "return_rate_pct", "monthly_units_sold"
]

df[num_cols] = df[num_cols].fillna(df[num_cols].median())

scalers = {
    "MinMax": MinMaxScaler(),
    "Standard": StandardScaler(),
    "Robust": RobustScaler(),
    "PowerTransform": PowerTransformer(method="yeo-johnson")
}

scaling_summary = []
for name, scaler in scalers.items():
    scaled_data = pd.DataFrame(scaler.fit_transform(df[num_cols]), columns=num_cols)
    scaling_summary.append({
        "Scaler": name,
        "Mean": round(scaled_data.mean().mean(), 4),
        "Std_Dev": round(scaled_data.std().mean(), 4),
        "Avg_Abs_Skewness": round(scaled_data.skew().abs().mean(), 4)
    })

summary_df = pd.DataFrame(scaling_summary)
print(summary_df)
```

```python
best_scaler = PowerTransformer(method="yeo-johnson")
df_scaled_num = pd.DataFrame(best_scaler.fit_transform(df[num_cols]), columns=[f"{c}_scaled" for c in num_cols])

print("Selected Scaler: PowerTransformer (Lowest Skewness)")
print(df_scaled_num.head())
```

## TAsk-2 Perform univariate analysis on numerical columns by implementing custom visualization functions for histograms, boxplots, KDE plots, and skewness/kurtosis calculations, ensuring insights into data distribution and potential transformations.

```python
stats_table = pd.DataFrame({
    "Mean": df[num_cols].mean().round(2),
    "Median": df[num_cols].median().round(2),
    "Std_Dev": df[num_cols].std().round(2),
    "Skewness": df[num_cols].skew().round(3),
    "Kurtosis": df[num_cols].kurt().round(3)
})

print("Numerical Distribution Statistics:")
print(stats_table)
```

```python
def plot_numerical_univariate(data, columns):
    for col in columns:
        fig, axes = plt.subplots(1, 3, figsize=(15, 4))
        
        sns.histplot(data[col], bins=25, ax=axes[0], color="#4C72B0")
        axes[0].set_title(f"Histogram: {col}")
        
        sns.boxplot(x=data[col], ax=axes[1], color="#55A868")
        axes[1].set_title(f"Boxplot: {col}")
        
        sns.kdeplot(data[col], fill=True, ax=axes[2], color="#C44E52")
        axes[2].set_title(f"KDE Plot: {col}")
        
        plt.tight_layout()
        plt.show()

plot_numerical_univariate(df, ["selling_price", "customer_age", "social_media_trend_score", "monthly_units_sold"])
```

## Task-3  Perform univariate analysis on categorical columns by creating frequency distributions, encoding high-cardinality categories dynamically, and visualizing trends using stacked bar charts, pie charts, and entropy-based category importance measures.

```python
cat_cols = [
    "brand_name", "division_name", "department_name", "class_name",
    "season", "occasion", "fabric_type", "pattern", "color",
    "size", "eco_friendly", "purchase_channel", "fit_feedback", "trend_demand_status"
]

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

entropy_list = []
for col in cat_cols:
    probs = df[col].value_counts(normalize=True)
    ent_val = entropy(probs, base=2)
    entropy_list.append({
        "Column": col,
        "Unique_Count": df[col].nunique(),
        "Entropy_Importance": round(ent_val, 3)
    })

entropy_df = pd.DataFrame(entropy_list).sort_values(by="Entropy_Importance", ascending=False).reset_index(drop=True)
print("Categorical Entropy Importance Table:")
print(entropy_df)
```

```python
high_card_cols = [col for col in cat_cols if df[col].nunique() > 10]
print("High-Cardinality Columns:", high_card_cols)

for col in high_card_cols:
    freq_encoding = df[col].value_counts(normalize=True)
    df[f"{col}_freq_encoded"] = df[col].map(freq_encoding).round(4)

print(df[[c for c in df.columns if "_freq_encoded" in c]].head())
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

dept_season = pd.crosstab(df["department_name"], df["season"], normalize="index") * 100
dept_season.plot(kind="bar", stacked=True, ax=axes[0], colormap="viridis")
axes[0].set_title("Stacked Bar: Department by Season (%)")
axes[0].set_ylabel("Percentage")
axes[0].legend(loc="upper right")

channel_counts = df["purchase_channel"].value_counts()
axes[1].pie(channel_counts, labels=channel_counts.index, autopct="%1.1f%%", startangle=90)
axes[1].set_title("Pie Chart: Purchase Channel Share")

sns.barplot(data=entropy_df, x="Entropy_Importance", y="Column", ax=axes[2], palette="magma")
axes[2].set_title("Category Importance by Entropy")

plt.tight_layout()
plt.show()
```
